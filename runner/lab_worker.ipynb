```
import os, pathlib, subprocess
tok_file = pathlib.Path('/content/drive/MyDrive/dcc_lab/.secrets/github_token.txt')
try:
    from google.colab import userdata
    os.environ['DCC_LAB_TOKEN'] = userdata.get('DCC_LAB_TOKEN')
except Exception:
    os.environ['DCC_LAB_TOKEN'] = tok_file.read_text().strip()
tok = os.environ['DCC_LAB_TOKEN']
if not pathlib.Path('/content/dcc-lab/.git').exists():
    subprocess.run(['git','clone','-q',f'https://x-access-token:{tok}@github.com/sooyahh/dcc-lab.git','/content/dcc-lab'], check=True)
subprocess.run(['git','-C','/content/dcc-lab','pull','-q'], check=True)
print('dcc-lab', subprocess.run(['git','-C','/content/dcc-lab','log','-1','--format=%h %s'],capture_output=True,text=True).stdout)
```
